# 01. Exploratory Data Analysis — NetSurfP-2.0 Benchmark Suite

This notebook explores the sequence characteristics, amino acid distributions, and secondary structure class balances across the **NetSurfP-2.0** training corpus and independent test sets (**CB513**, **TS115**, **CASP12**).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ssp.data.netsurfp import parse_netsurfp
from ssp.data.constants import Q8_CHARS, Q3_CHARS, AA_CHARS
from ssp.visualization.plotting import set_style, Q8_COLORS, OKABE_ITO

set_style()
%matplotlib inline
```

## 1. Load Processed Ragged Splits

```python
splits = {
    'train': parse_netsurfp('data/processed/train.npz'),
    'cb513': parse_netsurfp('data/processed/cb513.npz'),
    'ts115': parse_netsurfp('data/processed/ts115.npz'),
    'casp12': parse_netsurfp('data/processed/casp12.npz')
}

summary = []
for name, s in splits.items():
    lengths = s.lengths
    summary.append({
        'Split': name.upper(),
        'Sequences': len(s),
        'Residues': len(s.features),
        'Mean Length': np.mean(lengths),
        'Median Length': np.median(lengths),
        'Min Length': np.min(lengths),
        'Max Length': np.max(lengths),
    })
df_summary = pd.DataFrame(summary)
df_summary
```

## 2. DSSP Q8 Class Frequencies

```python
fig, ax = plt.subplots(figsize=(10, 4))
q8_tr = np.bincount(splits['train'].labels_q8, minlength=8) / len(splits['train'].labels_q8) * 100
q8_cb = np.bincount(splits['cb513'].labels_q8, minlength=8) / len(splits['cb513'].labels_q8) * 100

x = np.arange(8)
w = 0.35
ax.bar(x - w/2, q8_tr, width=w, label='Train', color='#0072B2')
ax.bar(x + w/2, q8_cb, width=w, label='CB513', color='#D55E00')
ax.set_xticks(x)
ax.set_xticklabels(Q8_CHARS)
ax.set_ylabel('Frequency (%)')
ax.set_title('Q8 Secondary Structure Distribution (Train vs CB513)')
ax.legend()
plt.show()
```
