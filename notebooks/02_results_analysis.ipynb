# 02. Benchmark Results & Error Analysis

This notebook evaluates model architectures (CNN, CNN-BiLSTM, BiLSTM-CRF) and input features (evolutionary HMM profiles vs one-hot sequences) across three gold-standard benchmarks:
- **CB513** (513 chains)
- **TS115** (115 high-resolution non-homologous chains)
- **CASP12** (21 blind competition targets)

```python
import json
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from ssp.data.constants import Q8_CHARS, Q3_CHARS
from ssp.visualization.plotting import set_style, OKABE_ITO

set_style()
%matplotlib inline
```

## 1. Load Evaluation Metrics

```python
df_bench = pd.read_csv('reports/results/metrics_benchmarks.csv')
df_bench
```

## 2. Pivot Table Summary (Q8 and Q3 Accuracy)

```python
pivot_q8 = df_bench.pivot(index='experiment', columns='benchmark', values='q8_acc')
pivot_q3 = df_bench.pivot(index='experiment', columns='benchmark', values='q3_acc')
print('Q8 Accuracy:')
display(pivot_q8)
print('\nQ3 Accuracy:')
display(pivot_q3)
```

## 3. Confusion Matrix Analysis (CB513)

```python
cm = np.load('reports/results/cm_q8_cb513.npy')
cm_norm = cm.astype(float) / np.maximum(cm.sum(axis=1, keepdims=True), 1) * 100
fig, ax = plt.subplots(figsize=(7, 6))
sns.heatmap(cm_norm, annot=True, fmt='.1f', cmap='Blues', xticklabels=Q8_CHARS, yticklabels=Q8_CHARS, ax=ax)
ax.set_title('CB513 DSSP Q8 Confusion Matrix (Recall %)')
ax.set_xlabel('Predicted')
ax.set_ylabel('True')
plt.show()
```
